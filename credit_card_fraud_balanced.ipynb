```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV 
from sklearn.preprocessing import StandardScaler  
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score,classification_report
```

```python
df=pd.read_csv(r"C:\Users\i7\Downloads\credit_card_fraud_balanced.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
la=LabelEncoder()
df["merchant_category"]=la.fit_transform(df["merchant_category"])
df["card_type"]=la.fit_transform(df["card_type"])
df["transaction_channel"]=la.fit_transform(df["transaction_channel"])
df["location"]=la.fit_transform(df["location"])
df["account_status"]=la.fit_transform(df["account_status"])
```

```python
df.head()
```

```python
df
```

```python
x=df.drop(columns=["is_fraud"])
y=df["is_fraud"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
df.info()
```

```python
# data scaling 
scalar=StandardScaler() 
x_train_scale=scalar.fit_transform(x_train)
x_test_scale=scalar.transform(x_test) 
```

```python
para_grid={"max_depth":np.arange(2,30,5),
          "n_estimators":np.arange(100,400,50)}
```

```python
tuning_model=RandomForestClassifier(random_state=42)
```

```python
grid_search=GridSearchCV(param_grid=para_grid,
                        estimator=tuning_model,
                        cv=5,
                        n_jobs=-1,
                        scoring='r2',
                        verbose=1) 

grid_search.fit(x_train,y_train)
```

```python
grid_search.best_params_
```

```python
model=grid_search.best_estimator_
```

```python
y_pred=model.predict(x_test)
precision_score(y_test,y_pred)
```

```python
y_pred=model.predict(x_test)
precision_score(y_test,y_pred)
```
